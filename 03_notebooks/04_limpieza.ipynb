```python
import polars as pl

archivo = pl.read_parquet("../02_data/raw/muestra_20k.parquet")
```

```python
limpia = archivo.with_columns(
    (pl.col("estado_contrato").str.to_lowercase()),
    (pl.col("sector").str.to_lowercase()),
    (pl.col("ciudad").str.to_lowercase()),
    (pl.col("departamento").str.to_lowercase()),
    (pl.col("rama").str.to_lowercase()),
    (pl.col("orden").str.to_lowercase()),
    (pl.col("modalidad_de_contratacion").str.to_lowercase()),
    (pl.col("destino_gasto").str.to_lowercase()),
    (pl.col("entidad_centralizada").str.to_lowercase()),
    (pl.col("liquidaci_n").str.to_lowercase()),
    (pl.col("es_pyme").str.to_lowercase()),
    (pl.col("es_grupo").str.to_lowercase()),
    (pl.col("origen_de_los_recursos").str.to_lowercase()),
    (pl.col("justificacion_modalidad_de").str.to_lowercase()),
    (pl.col("tipodocproveedor").str.to_lowercase()),
    (pl.col("habilita_pago_adelantado").str.to_lowercase()),
    (pl.col("tipo_de_contrato").str.to_lowercase()),


)

```

```python
print(limpia["estado_contrato"].value_counts(sort=True))
print(limpia["sector"].value_counts(sort=True))
print(limpia["ciudad"].value_counts(sort=True))
print(limpia["departamento"].value_counts(sort=True))
print(limpia["rama"].value_counts(sort=True))
print(limpia["orden"].value_counts(sort=True))
print(limpia["modalidad_de_contratacion"].value_counts(sort=True))
print(limpia["justificacion_modalidad_de"].value_counts(sort=True))
print(limpia["destino_gasto"].value_counts(sort=True))
print(limpia["entidad_centralizada"].value_counts(sort=True))
print(limpia["liquidaci_n"].value_counts(sort=True))
print(limpia["es_pyme"].value_counts(sort=True))
print(limpia["es_grupo"].value_counts(sort=True))
print(limpia["origen_de_los_recursos"].value_counts(sort=True))
print(limpia["habilita_pago_adelantado"].value_counts(sort=True))
print(limpia["tipo_de_contrato"])
```
